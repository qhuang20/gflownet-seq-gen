# Explore the verified toy results

Runs without the private `gfn` package. These are recorded results from the 4bp CPU notebook, not a new training run.

TB/DB and FL-DB used different reward functions, so their hit rates are not a controlled comparison. No biological function has been validated.

```python
import json
from pathlib import Path

report = json.loads(Path("examples/toy_results.json").read_text())
print(f"Verified: {report['verified_on']}; seed: {report['seed']}; episodes per run: {report['episodes_per_run']:,}")
for name, rate in report["target_hit_rate_percent"].items():
    print(f"{name}: {rate:.2f}% exact target hits during training")
```

```python
import matplotlib.pyplot as plt

rates = report["target_hit_rate_percent"]
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(list(rates), list(rates.values()), color=["#34778a", "#689f87", "#cc9e50"])
ax.set_ylabel("Exact target hits during training (%)")
ax.set_title("4bp toy runs · different rewards across objectives")
ax.set_ylim(0, 12)
for i, rate in enumerate(rates.values()):
    ax.text(i, rate + 0.2, f"{rate:.2f}%", ha="center")
fig.tight_layout()
plt.show()
```
